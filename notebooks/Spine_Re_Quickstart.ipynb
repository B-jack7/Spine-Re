## 2. 运行当前阶段
smoke：2 轮，8/4/4 张图。train：完整数据，每次到下一个断点暂停。续跑同一实验请保持配置不变，再次运行本单元格。

点击停止会结束训练子进程；当前轮次未完成的进度会丢弃。续跑从最近完整断点开始。网络断开不等于训练停止；内核被强制关闭时无法运行清理逻辑。

```python
runner_source = '"""Stream a training command while owning its lifetime (standard library only)."""\n\nimport subprocess\nfrom pathlib import Path\n\n\ndef run_logged(command, log_path, emit=None, stop_timeout=10):\n    """Append combined output; stop and reap this child if the cell is interrupted.\n\n    This stops the current epoch, not saves it. Resume from the last completed\n    checkpoint. Runtime deletion or a killed notebook kernel cannot run cleanup.\n    The trainer currently uses num_workers=0 and creates no child workers.\n    """\n    if emit is None:\n        emit = lambda line: print(line, end="", flush=True)\n    with Path(log_path).open("a", encoding="utf-8") as log:\n        proc = subprocess.Popen(\n            command, stdout=subprocess.PIPE, stderr=subprocess.STDOUT,\n            text=True, encoding="utf-8", errors="replace", bufsize=1,\n        )\n        try:\n            for line in proc.stdout:\n                log.write(line)\n                log.flush()\n                emit(line)\n            returncode = proc.wait()\n            if returncode:\n                raise subprocess.CalledProcessError(returncode, command)\n        finally:\n            try:\n                if proc.poll() is None:\n                    proc.terminate()\n                    try:\n                        proc.wait(timeout=stop_timeout)\n                    except subprocess.TimeoutExpired:\n                        proc.kill()\n                        proc.wait()\n            finally:\n                proc.stdout.close()\n'
import runpy
runner_path=REPO/'scripts'/'notebook_process.py'
runner_path.parent.mkdir(exist_ok=True)
runner_path.write_text(runner_source,encoding='utf-8')
run_logged=runpy.run_path(str(runner_path))['run_logged']
cmd=[sys.executable,'-u',str(code_dir/'train.py'),'--repo',str(REPO),'--output',str(RUN_DIR),'--model',MODEL,'--label-policy',LABEL_POLICY]
if MODE=='smoke':
    cmd += ['--epochs','2','--train-limit','8','--val-limit','4','--test-limit','4']
else:
    cmd += ['--epochs',str(EPOCHS),'--pause-every',str(PAUSE_EVERY)]
if (RUN_DIR/'last.pt').exists():cmd.append('--resume')
try:
    run_logged(cmd,RUN_DIR/'console.log')
except KeyboardInterrupt:
    print('训练已停止。当前轮次未完成的进度不会保存；再次运行将从最近完整断点续跑。')
    raise
if (RUN_DIR/'test_metrics.json').exists():
    print((RUN_DIR/'test_metrics.json').read_text())
    if MODE=='smoke':print('SMOKE PASSED：仅验证流程，小样本分数不代表模型性能。')
else:
    last=json.loads((RUN_DIR/'history.jsonl').read_text().splitlines()[-1])
    print('阶段暂停，已完成轮数：',last['epoch'])
    print('验证结果：',json.dumps(last['validation'],indent=2))
```
